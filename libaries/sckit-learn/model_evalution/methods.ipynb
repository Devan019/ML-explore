```python
y_true = [0,1,2,1,0,2,1,0,2,1,0,1,2,2,0,0,1,2,1,0,2,1,0,2,1]

y_pred = [0,1,1,1,0,2,0,0,2,1,0,2,2,1,0,0,1,2,1,0,2,0,0,2,1]
```

classification method

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
# accuracy

accuracy = accuracy_score(y_true, y_pred) # 80%

#precision
# for imbalanced - weighted, equally - macro
pricsion = precision_score(y_true, y_pred, average="macro")

#recall
recall = recall_score(y_true, y_pred, average="macro")

#f1
f1 = f1_score(y_true, y_pred, average="macro")

print(accuracy, pricsion, recall, f1)
```

confusion matrix

```python
from sklearn.metrics import confusion_matrix
mat = confusion_matrix(y_pred=y_pred, y_true=y_true)

mat
```
