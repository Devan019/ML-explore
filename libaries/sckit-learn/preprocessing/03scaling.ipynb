```python
import pandas as pd
data = pd.read_csv("after02.csv")
ndata = data.drop(columns=["name"])
ndata.head(1)
```

type-1 standard scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()

scaled = sc.fit_transform(ndata)

pd.DataFrame(scaled, columns=ndata.columns)
```

type - 2 MinMax scaling

```python
from sklearn.preprocessing import MinMaxScaler
mc = MinMaxScaler()

mscaled = mc.fit_transform(ndata)

# i like this scale :) - so i store
ndata = pd.DataFrame(mscaled, columns=ndata.columns)
ndata
```

```python
# export data
ndata.to_csv("after03.csv", index=False)
```
