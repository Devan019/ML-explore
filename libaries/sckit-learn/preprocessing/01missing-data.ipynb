fill or drop data on missing value

```python
import pandas as pd
data = pd.read_csv("data.csv")
data
```

```python
#proper data handing - name null remove, score - mean, city- mod, placment - 0

# #name null remov
data = data.dropna(subset="name").copy()

# #placment - 0
data["placement"] = data["placement"].fillna(0) 

#city - mode
mfreq = data["city"].mode()[0]
data["city"] = data["city"].fillna(mfreq)

#score - mean
data["score"] = data["score"].fillna(data["score"].mean())

data
```

```python
#export data as csv
data.to_csv("after01.csv", index=False)
```
