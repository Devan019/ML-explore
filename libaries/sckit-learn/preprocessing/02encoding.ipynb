```python
import pandas as pd
data = pd.read_csv("after01.csv")
data.info()
```

type 1 - label encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

#hear we only do on true/false - 0/1
data["placement"] = le.fit_transform(data["placement"])
data
```

type-2 one-hot method

```python

data = pd.get_dummies(data=data, columns=["city"], dtype=int)
data
```

```python
# export data
data.to_csv("after02.csv", index=False)
```
