```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [10,20,15,25]

plt.plot(x,y)
plt.show()
```

```python
x = ["Mon", "Tue", "Wed", "Thu", "Fri"] 
y = [10,15,7,20,13]

plt.plot(x,y, color="green", linestyle="--", linewidth=1.2, marker = "o", label = "bakery data")
plt.title("Bakery sales this week!")
plt.xlabel("Day of the weak")
plt.ylabel("No of item to sold")
plt.legend(loc="upper left", fontsize=10)
plt.grid(color = "gray", linestyle=":")
plt.ylim(0,30)
# plt.xlim(0,6)
plt.xticks(["Mon", "Tue", "Wed", "Thu", "Fri"], ["D1", "D2", "D3","D4", "D5"])
plt.show()
```
