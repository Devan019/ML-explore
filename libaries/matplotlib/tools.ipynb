bars, pie charts, histograms

```python
import matplotlib.pylab as plt
```

bar

```python
product = ["A", "B", "C", "D"]
sales =[1000, 1500, 800, 1200]

plt.bar(product, sales, 0.5, label = "2025 sales", color="orange")
plt.xlabel("Produts")
plt.ylabel("sales")
plt.title("Product sales comparison")
plt.legend()
plt.ylim(0,1600)
plt.show()
```

barh

```python
plt.barh(product, sales, 0.5, label = "2025 sales", color="orange")
plt.xlabel("Produts")
plt.ylabel("sales")
plt.title("Product sales comparison")
plt.legend()
plt.xlim(0,1600)
plt.show()
```
