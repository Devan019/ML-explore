14. Convertors

```python
#example - rename a row full-form to short name like RCB, MI
def rename_conveter(name: str):
  name_arr = name.split(" ")
  new_name = ""
  for n in name_arr:
    new_name += n[0]
  return new_name


pd.read_csv("matches.csv", converters={
  "team1" : rename_conveter,
  "team2" : rename_conveter,
  "toss_winner": rename_conveter
}).head()
```
