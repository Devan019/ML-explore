Power transformer

```python
from sklearn.preprocessing import PowerTransformer
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb

df = pd.read_csv("concrete_data.csv")
df.head(3)
```

```python
pt = PowerTransformer(method="box-cox")
for i in range(8):

  col = df.iloc[:, i:i+1]

  plt.subplot(1,2,1)
  sb.histplot(col, kde=True)

  plt.subplot(1,2,2)
  sb.histplot(pt.fit_transform(col+0.000000001), kde=True)
  plt.show()
```

```python
pt2 = PowerTransformer(method="yeo-johnson")
for i in range(8):

  col = df.iloc[:, i:i+1]

  plt.subplot(1,2,1)
  sb.histplot(col, kde=True)

  plt.subplot(1,2,2)
  sb.histplot(pt.fit_transform(col+0.000000001), kde=True)
  plt.show()
```

```python
pt.fit_transform(df.iloc[:,0:8]+0.00001)
pt2.fit_transform(df.iloc[:,0:8])

pd.DataFrame({'cols':df.iloc[:,0:8].columns,'box_cox_lambdas':pt.lambdas_,'Yeo_Johnson_lambdas':pt2.lambdas_})
```
