Function Transformer

```python
import seaborn as sb
import pandas as pd
from sklearn.preprocessing import FunctionTransformer
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("../../train.csv")
df.head(3)
```

```python
import seaborn as sb
import matplotlib.pyplot as plt
from sklearn.preprocessing import FunctionTransformer
import numpy as np

fare = df["Fare"].dropna()

def transformer(fun):
    ft = FunctionTransformer(fun)
    fare_t = ft.fit_transform(fare.values.reshape(-1, 1)).flatten()

    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    sb.histplot(fare, kde=True)
    plt.title("Original Fare")

    plt.subplot(1, 2, 2)
    sb.histplot(fare_t, kde=True)
    plt.title(f"Transformed with {fun.__name__}")

    plt.show()

transformer(np.sin)
transformer(np.sqrt)
transformer(np.square)
transformer(np.cos)
transformer(np.log1p)
```
