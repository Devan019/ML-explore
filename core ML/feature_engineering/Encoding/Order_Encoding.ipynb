### Encoding Categorical Data

#### Lable Encoding - Odinal Data
ordered data

```python
import pandas as pd
df = pd.read_csv("./customer.csv")
df.sample(3)
```

```python
df["gender"].unique(),df["review"].unique(),df["education"].unique(),df["purchased"].unique()
```

```python
#here review, education and purchased are order data

from sklearn.model_selection import train_test_split

X = df.iloc[:,2:4]
Y = df.iloc[:,4:]

X_train: pd.DataFrame
X_test: pd.DataFrame
Y_train: pd.Series
Y_test: pd.Series

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.33, random_state=45)
```

```python
X_train.head(5)
```

```python
#do encoding
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories= [ ['Poor', 'Average', 'Good'] , ['School', 'UG', 'PG']] )
#here categories belong to set order lower to hig for all cols

oe.fit(X_train)

X_train_tranformed = oe.transform(X_train)
X_test_tranformed = oe.transform(X_test)

X_train_tranformed_df = pd.DataFrame(X_train_tranformed, columns=X_train.columns)
X_test_tranformed_df = pd.DataFrame(X_test_tranformed, columns=X_test.columns)

X_train_tranformed_df.head(5) #order set :)
```

```python
# for output use LabelEncoding 
from sklearn.preprocessing import LabelEncoder
le =LabelEncoder()
le.fit(Y_train)

Y_train_tranformed = le.transform(Y_train)
Y_test_tranformed = le.transform(Y_test)

Y_train_tranformed_df = pd.DataFrame(Y_train_tranformed, columns=Y_train.columns)
Y_test_tranformed_df = pd.DataFrame(Y_test_tranformed, columns=Y_test.columns)

Y_train_tranformed_df.head(5) #order set :)
```
