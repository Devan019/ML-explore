```python
import pandas as pd
import numpy as np
df = pd.read_csv("./covid.csv")

from sklearn.compose import ColumnTransformer

df.sample(5)
```

```python
df["city"].unique()
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer #fill missing value which is present in fever
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, 
```

```python
transforme = ColumnTransformer(
  transformers=[
    ("t1", SimpleImputer(), ["fever"]), #first transformation like wise - (name, work, col)
    ("t2", OneHotEncoder(drop="first", dtype=np.int8, sparse_output=False), ["gender", "city"]),
    ("t3", OrdinalEncoder(categories=[["Mild", "Strong"]], dtype=np.int8), ["cough"]),
  ],
  remainder="passthrough" #means other cols are remain same not chanage it ex-  age
)
```

```python
df_transformed = transforme.fit_transform(df)
df_transformed.shape
#age - 1, gender - 1, fever-1, cough - 1, city - 3, has _covid - 1 => here col - # of col => total = 8
```
